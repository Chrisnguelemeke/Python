```python
%%writefile mp_times.py
import multiprocessing
import random
import time
from datetime import datetime

def wait_and_print(name):
    seconds = random.random()
    time.sleep(seconds)
    print(f'{name} waited {seconds:.2f} sec, time is {datetime.now().time()}')

if __name__ == '__main__':
    processes = []
    for n in range(1, 4):
        p = multiprocessing.Process(target=wait_and_print, args=(f'Process {n}',))
        processes.append(p)
        p.start()
    for p in processes:
        p.join()
```
