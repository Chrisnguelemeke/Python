```python
from datetime import date

now = date.today()
now_str = now.isoformat()

with open('today.txt', 'wt') as output:
    print(now_str, file=output)

now_str
```

```python
with open('today.txt', 'rt') as input_file:
    today_string = input_file.read()

today_string
```
