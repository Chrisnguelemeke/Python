```python
# Name: Chris-Bradley Harry Nguele Meke
# File Name: student_gpa_checker.py
# Description: This program accepts student information and determines
#              whether each student qualifies for the Dean's List or Honor Roll.

while True:
    last_name = input("Enter student's last name (ZZZ to quit): ").strip()

    if last_name.upper() == "ZZZ":
        break

    first_name = input("Enter student's first name: ").strip()
    gpa = float(input("Enter student's GPA: "))

    if gpa >= 3.5:
        print(f"{first_name} {last_name} has made the Dean's List.")

    if gpa >= 3.25:
        print(f"{first_name} {last_name} has made the Honor Roll.")

    print()
```
